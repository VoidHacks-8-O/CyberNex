# 🏆 Kaggle Grandmaster End-to-End Competitive ML Pipeline
### Optimized for 200,000+ Row Tabular Datasets

**Architecture Highlights:**
1. **Memory Downcasting Engine**: Automatic conversion of `int64` / `float64` to `int16/32` & `float32` preventing OOM on 200k+ rows.
2. **Automated Feature Engineering**: Frequency encoding, numerical interaction terms (sums, diffs, products, ratios), and categorical aggregations.
3. **GBDT Trifecta Ensemble**: LightGBM + XGBoost + CatBoost trained with 5-Fold Stratified Cross-Validation.
4. **Optimal SLSQP Blending**: Solves for optimal continuous ensemble weights directly on Out-Of-Fold (OOF) predictions.
5. **Optuna Tuning & SHAP Explainability**: Built-in Bayesian hyperparameter search and TreeExplainer interpretation.
6. **Production Artifacts**: `submission.csv`, feature importance barplots, and `ensemble_pipeline.joblib` bundle.
